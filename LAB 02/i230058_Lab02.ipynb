```python
##Task 03
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style="whitegrid")
titanic.isnull().sum()
titanic['age'] = titanic['age'].fillna(titanic['age'].mean())
titanic['embarked'] = titanic['embarked'].fillna(titanic['embarked'].mode()[0])
titanic = titanic.drop(columns=['deck'])
titanic.isnull().sum()

```

```python
##Task 04
def remove_duplicates(df, name):
    before = df.shape
    df_clean = df.drop_duplicates()
    after = df_clean.shape
    print(f"{name}: {before} → {after}")
    return df_clean

iris = remove_duplicates(iris, "IRIS")
titanic = remove_duplicates(titanic, "TITANIC")
tips = remove_duplicates(tips, "TIPS")

```

```python
##Task 05
iris.groupby("species").agg(["mean", "std"])

titanic.groupby("survived")[["age", "fare"]].mean()

tips[["total_bill", "tip"]].agg(["mean", "median"])

```

```python
##Task 07
def iqr_outliers(df, col):
    Q1 = 
df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return lower, upper, outliers.shape[0]

for col in ['total_bill', 'tip']:
    l, u, c = iqr_outliers(tips, col)
    print(col, "Outliers:", c)
tips.boxplot(column=['total_bill', 'tip'])
plt.show()

```

```python
##Task 08
def remove_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    return df[(df[col] >= Q1 - 1.5*IQR) & (df[col] <= Q3 + 1.5*IQR)]

before = tips.shape
tips_clean = remove_iqr(tips, 'total_bill')
tips_clean = remove_iqr(tips_clean, 'tip')
after = tips_clean.shape

before, after
```

```python
##Task 09
comparison = pd.concat([
    tips[['total_bill','tip']].agg(['mean','median','std']),
    tips_clean[['total_bill','tip']].agg(['mean','median','std'])
], axis=1, keys=['Before', 'After'])

comparison
```
