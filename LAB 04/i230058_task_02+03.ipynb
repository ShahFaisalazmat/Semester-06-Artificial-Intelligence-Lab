```python
# City map as adjacency list
graph = {
    'A': [('B', 4), ('C', 3)],
    'B': [('D', 5), ('E', 6)],
    'C': [('E', 2)],
    'D': [('F', 4)],
    'E': [('F', 3)],
    'F': []
}

start = 'A'      # Depot
goal = 'F'       # Customer

# Find all reachable states
def get_reachable():
    visited = set()
    queue = [start]
    
    while queue:
        node = queue.pop(0)
        if node not in visited:
            visited.add(node)
            for neighbor, _ in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return visited

# Find best path
best_path = None
best_cost = float('inf')

def dfs(current, path, cost, visited):
    global best_path, best_cost
    path.append(current)
    visited.add(current)
    if current == goal:
        if cost < best_cost:
            best_path = path.copy()
            best_cost = cost
    else:
        for neighbor, weight in graph[current]:
            if neighbor not in visited:
                dfs(neighbor, path.copy(), cost + weight, visited.copy())
dfs(start, [], 0, set())
print(f"Start: {start}")
print(f"Goal: {goal}")
print(f"\nReachable states: {sorted(get_reachable())}")
print(f"\nBest path: {' → '.join(best_path)}")
print(f"Total cost: {best_cost}")
print("\nActions:")
for i in range(len(best_path)-1):
    for neighbor, weight in graph[best_path[i]]:
        if neighbor == best_path[i+1]:
            print(f"MoveTo({neighbor}) [cost: {weight}]")
            break
```

```python
goal_state = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)
initial_state = (
    (1, 2, 3),
    (4, 0, 6),
    (7, 5, 8)
)
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
    return None
def get_moves(state):
    moves = []
    i, j = find_blank(state)
    state_list = [list(row) for row in state]
    for move, di, dj in [('Up', -1, 0), ('Down', 1, 0), ('Left', 0, -1), ('Right', 0, 1)]:
        new_i, new_j = i + di, j + dj
        
        if 0 <= new_i < 3 and 0 <= new_j < 3:
            # Swap blank with tile
            state_list[i][j], state_list[new_i][new_j] = state_list[new_i][new_j], state_list[i][j]
            new_state = tuple(tuple(row) for row in state_list)
            moves.append((move, new_state))
            
            # Swap back for next move
            state_list[i][j], state_list[new_i][new_j] = state_list[new_i][new_j], state_list[i][j]
    return moves
def is_goal(state):
    return state == goal_state
def print_state(state):
    for row in state:
        print(' '.join(str(x) if x != 0 else '_' for x in row))
    print()
def generate_state_space(max_depth=10):
    states = {initial_state: (None, None, 0)}  # state: (parent, move, depth)
    frontier = [(initial_state, 0)]  # (state, depth)
    while frontier:
        current_state, depth = frontier.pop(0)
        if depth >= max_depth:
            continue
        for move, new_state in get_moves(current_state):
            if new_state not in states:
                states[new_state] = (current_state, move, depth + 1)
                frontier.append((new_state, depth + 1))
                if is_goal(new_state):
                    return states, new_state
    return states, None
def find_path(states, goal_state):
    if not goal_state:
        return [] 
    path = []
    current = goal_state
    while current:
        parent, move, depth = states[current]
        if parent:
            path.append((move, current))
        current = parent
    path.reverse()
    return path
print("Initial State:")
print_state(initial_state)
print("Goal State:")
print_state(goal_state)
print("Generating state space (max depth: 10)...")
states, goal = generate_state_space(10)
print(f"\nTotal states generated: {len(states)}")
path = find_path(states, goal)
if path:
    print(f"\nSolution found! Path cost: {len(path)}")
    print("\nSolution path:")
    print("Step 0 (Initial):")
    print_state(initial_state)
    for step, (move, state) in enumerate(path, 1):
        print(f"Step {step}: Move {move}")
        print_state(state)
else:
    print("\nNo solution found within depth limit of 10.")
```
