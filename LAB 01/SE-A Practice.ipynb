## Dictionaries for State Representation


```python
heuristic = {
    "A": 10,
    "B": 5,
    "C": 2
}

heuristic["B"]
```

```python
heuristic["D"] = 1
heuristic["A"] = 8
heuristic
```

```python
for state, h in heuristic.items():
    print(state, h)
```

```python
heuristic.get("E", "Not Found")
```

```python
students = {}

for i in range(0, 2):
    name = input("Enter your name: ") #user input
    reg = input("Enter registration: ")
    gpa = float(input("Enter gpa: "))

    students[i] = {
        "Student_Name": name,
        "Student_Category": reg,
        "GPA": gpa
    }
```

```python
students[1]
```

```python
sorted_students = sorted(students.items(), key=lambda item: item[1]["GPA"])
print(sorted_students)
```

```python
heuristic = lambda x: x * 2
heuristic(5)
```

# Reduce

```python
from functools import reduce

n = [1, 2, 3, 4, 5, 6]

res = reduce(lambda x, y: x + y, n)

print(res)
```

## Object-Oriented Programming for AI


```python
class Agent:
    def __init__(self, name):
        self.name = name

    def act(self, percept):
        return f"{self.name} received {percept}"
```

```python
agent1 = Agent("SimpleAgent")
agent1.act("clean")
```
